```python
import pandas as pd

df = pd.read_excel(
    "../Data_understanding/Project_106_Unclean_Energy_Dataset_5000.xlsx",
    sheet_name="Unclean_Data"
)

df.head()
```

```python
df_clean = df.copy()
```

```python
df_clean.isna().sum()
```

```python
df_clean.duplicated().sum()
```

```python
df_clean = df_clean.drop_duplicates()
```

```python
df_clean.duplicated().sum()
```

```python
df_clean["State"] = df_clean["State"].str.title()
```

```python
df_clean["Sector"] = df_clean["Sector"].str.title()
```

Date_Time clean

```python
df_clean["Date_Time"] = pd.to_datetime(df_clean["Date_Time"], errors="coerce")
```

```python
df_clean["Date_Time"].isna().sum()
```

```python
df_clean = df_clean.dropna(subset=["Date_Time"])
```

Missing values-a handle

```python
df_clean["State"].isna().sum()
```

```python
df_clean["State"] = df_clean["State"].fillna(df_clean["State"].mode()[0])
```

```python
df_clean["Sector"] = df_clean["Sector"].fillna(df_clean["Sector"].mode()[0])
```

```python
df_clean["Consumption_MWh"] = df_clean["Consumption_MWh"].fillna(
    df_clean["Consumption_MWh"].median()
)
```

```python
df_clean["Peak_Demand_MW"] = df_clean["Peak_Demand_MW"].fillna(
    df_clean["Peak_Demand_MW"].median()
)
```

```python
df_clean["Generation_MWh"] = df_clean["Generation_MWh"].fillna(
    df_clean["Generation_MWh"].median()
)
```

```python
df_clean["Electricity_Price_USD_per_MWh"] = df_clean["Electricity_Price_USD_per_MWh"].fillna(
    df_clean["Electricity_Price_USD_per_MWh"].median()
)
```

```python
df_clean["Renewable_Share_Pct"] = df_clean["Renewable_Share_Pct"].fillna(
    df_clean["Renewable_Share_Pct"].median()
)
```

```python
df_clean["Weather_Index"] = df_clean["Weather_Index"].fillna(
    df_clean["Weather_Index"].median()
)
```

```python
df_clean["Peak_Flag"].isna().sum()
```

invalid values

```python
df_clean = df_clean[df_clean["Consumption_MWh"] >= 0]
```

```python
df_clean = df_clean[df_clean["Generation_MWh"] >= 0]
```

```python
df_clean = df_clean[
    (df_clean["Renewable_Share_Pct"] >= 0) &
    (df_clean["Renewable_Share_Pct"] <= 100)
]
```
